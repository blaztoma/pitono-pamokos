# `range()` function

We will use the `range()` function when we want to generate a certain sequence of numbers.

The `range()` function is often used with the `for` loop.

## Syntax

### 1st option

```
range(end)
```

- from 0 (included)
- to `end` (excluded)

### 2nd option

```
range(start, end)
```

- from `start` (included)
- to `end` (excluded)

### 3rd option

```
range(start, end, step)
```

- from `start` (included)
- to `end` (excluded)
- taking the specified step `step` (using a positive value - will generate numbers in increasing order, using a negative value - will generate numbers in decreasing order)

### A few syntax examples

````python
range(7)        # from 0 to 7 (numbers 0, 1, 2, 3, 4, 5, 6)
range(2, 8)     # from 2 to 8 (numbers 2, 3, 4, 5, 6, 7)
range(1, 6, 2)  # odd numbers from 1 to 6 (1, 3, 5)
range(7, 0, -1) # from 7 to 1 (numbers 7, 6, 5, 4, 3, 2, 1)
````

## Examples

The `range()` function is unique in that it creates a `range` object. This object is not a list of numbers.

````python
print( range(7) )
````

To get a list of numbers from a `range` object, we need to use the `list()` function.

````python
numbers = list( range(6) )
print(numbers)
````

You've probably already noticed that `range()` does not include the last specified number. Why is that?

````python
print( list( range(10) ) )
````

Using `range` we can specify the start.

````python
numbers = list( range(2, 7) )
print(numbers)
````

Another example.

````python
nums = list( range(10, 15) )
print(nums)
````

This example includes a `step` with a value of 2. In this case, we will print odd numbers.

````python
every_second = list( range(5, 15, 2) )
print(every_second)
````

Another example with a `step` value of 5. Note that it does not include the last value (`50`).

````python
numbers = list( range(5, 50, 5) )
print(numbers)
````

Now that we know the nuances of using the `range()` function, let's practice. Creating a list [1, 50].

````python
numbers = list( range(1, 51) )
print(numbers)
````

And here - creating a list of odd numbers [1, 29].

````python
odd_numbers = list( range(1, 30, 2) )
print(odd_numbers)
````

In this example, we will take every sixth number.

````python
numbers = list( range(5, 87, 6) )
print(numbers)
````

What we haven't tried yet is creating a list of numbers in decreasing order. For that, we need to specify a negative `step` value. Note that the last value is not included in this case either.

````python
numbers = list( range(10, 0, -1) )
print(numbers)
````

Another example where the `step` decreases by five units.

````python
numbers = list( range(50, 0, -5) )
print(numbers)
````

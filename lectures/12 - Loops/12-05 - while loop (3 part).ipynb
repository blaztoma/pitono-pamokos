# `while` Loop (Part 3)

## Examples

We will ask to enter a name until at least some text is entered.

```python
name = None

while name == '' or name == None:
    name = input('Enter your name: ')

print('Entered name:', name)
```

We will supplement this program with a message that will be displayed if the name was not entered. For this purpose, we compare the name with an empty string.

```python
name = None

while name == '' or name == None:
    name = input('Enter your name: ')
    if name == '':
        print('Incorrectly entered name, try again')

print('Entered name:', name)
```

In this program, we check the name as a boolean variable, which will automatically have the value `False` if it is empty.

```python
name = None

while name == '' or name == None:
    name = input('Enter your name: ')
    if not name:
        print('Incorrectly entered name, try again')

print('Entered name:', name)
```

Here we move the check into the loop condition. Because of this, we organize the first name input outside the loop.

```python
name = input('Enter your name: ')

while not name:
    print('Incorrectly entered name, try again.')
    name = input('Enter your name: ')

print('Entered name:', name)
```

We can do the same with entered numbers. In this program, we will input a number until it falls within the specified range.

```python
start, end = 1, 100

number = 0

while number < start or number > end:
    number = int(input('Enter a number between [1-100]: '))

print('Entered number:', number)
```

We add information about the fact that the number does not fall into the interval.

```python
start, end = 1, 100

number = 0

while number < start or number > end:
    number = int(input('Enter a number between [1-100]: '))
    if number < start or number > end:
        print('Number is outside the specified range!')

print('Entered number:', number)
```

In this program, we organize an infinite loop for input and escape from it.

```python
start, end = 1, 100

number = 0

while True:
    number = int(input('Enter a number between [1-100]: '))
    if number < start or number > end:
        print('Number is outside the specified range!')
    else:
        break

print('Entered number:', number)
```

When working with numbers and input, we should always check if the entered value is a number.

```python
input_value = input('Enter a number: ')

while not input_value.isdigit():
    input_value = input('You entered not a number, try again: ')

number = int(input_value)
print('Entered number:', number)
print('Its type:', type(number))
print('Obtained from input:', input_value, type(input_value))
```

Input of two numbers in the form presented above.

```python
first = input('Enter the first number: ')

while not first.isdigit():
    first = input('Enter the first number: ')

second = input('Enter the second number: ')

while not second.isdigit():
    second = input('Enter the second number: ')

first = int(first)
second = int(second)

print(f'{first} + {second} = {first + second}')
print(f'{first} - {second} = {first - second}')
print(f'{first} * {second} = {first * second}')
print(f'{first} / {second} = {first / second}')
```

We shorten the program by assigning empty values to input variables. The loop condition will not be satisfied and therefore it will ask to enter a number.

```python
first, second = '', ''

while not first.isdigit():
    first = input('Enter the first number: ')

while not second.isdigit():
    second = input('Enter the second number: ')

first = int(first)
second = int(second)

print(f'{first} + {second} = {first + second}')
print(f'{first} - {second} = {first - second}')
print(f'{first} * {second} = {first * second}')
print(f'{first} / {second} = {first / second}')
```

To check if any conversion (whatever it may be) was successful, we can use exception handling. For this we use a `try` ... `except` block.

```python
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("This is not a number!")

print('Entered number:', number)
```

Exception handling used inside a loop.

```python
repeat = True

while repeat:
    try:
        number = int(input("Enter a number: "))
        repeat = False
    except ValueError:
        print("This is not a number!")

print('Entered number:', number)
```

This program organizes the input of many numbers and calculates their sum.

```python
sum_value = 0
repeat = True
input_value = ''

while repeat:
    while not input_value.isdigit():
        input_value = input('Enter a number: ')

    number = int(input_value)
    print('Entered number:', number)
    sum_value += number

    input_value = input('Do you want to continue? y/n ')
    if input_value != 't':
        repeat = False

print('Sum of all entered numbers:', sum_value)
```

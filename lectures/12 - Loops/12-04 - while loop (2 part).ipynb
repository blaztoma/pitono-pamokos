# `while` Loop (Part 2)

## Examples

We repeat the operation (in this case - the sum of two numbers) until the user confirms the input (enters a letter other than `t`).

````python
repeat = 't'

while repeat == 't':
    print("Enter two numbers (separated by enter):")
    first = int(input())
    second = int(input())
    print(f'{first} + {second} = {first + second}')
    repeat = input('Do you want to repeat? (y/n): ')

print('End')
````

We can do the same using an infinite loop and escape from it using the `break` operation.

````python
while True:
    print("Enter two numbers (separated by enter):")
    first = int(input())
    second = int(input())
    print(f'{first} + {second} = {first + second}')
    repeat = input('Do you want to repeat? (y/n): ')
    if repeat != 't':
        print('No more repetitions')
        break

print('End')
````

Try to guess a random number generated by the computer. Pay attention to the `while` loop and its condition.

````python
import random

secret = random.randint(1, 10)
guess = None

while secret != guess:
    guess = int(input('Guess a number between 1 and 10: '))

print('You guessed it!')
print('The secret number was:', secret)
````

Let's improve the program a bit so it tells you how well you guessed.

````python
import random

secret = random.randint(1, 10)
guess = None

while secret != guess:
    guess = int(input('Guess a number between 1 and 10: '))
    if secret != guess:
        print('Sorry, try again')
    else:
        print('You won!')
````

Let's improve the program further by telling the user whether the guessed number is larger or smaller.

````python
import random

secret = random.randint(1, 10)
guess = None

while secret != guess:
    guess = int(input('Guess a number between 1 and 10: '))
    if secret > guess:
        print('Try a larger number')
    elif secret < guess:
        print('Try a smaller number')
    else:
        print('You guessed it!')
````

We will use a loop to calculate the sum and average. We will input numbers as long as the user wants, and then calculate the average.

````python
grades_sum = 0
grades_count = 0

continue_input = 'y'

while continue_input == 'y':
    grade = int(input('Enter a grade: '))
    grades_sum += grade
    grades_count += 1
    continue_input = input('Do you want to enter more? (y/n) ')

average = round(grades_sum / grades_count, 1)
print('Average of entered grades:', average)
````

Another implementation of the same algorithm. This time we enter 0 to finish. This saves us one input operation.

````python
grades_sum = 0
grades_count = 0

grade = -1

print('Enter as many grades as you want (separated by enter)')
print('To finish, enter 0')

while grade != 0:
    grade = int(input('Enter a grade: '))
    if grade != 0:
        grades_sum += grade
        grades_count += 1

average = round(grades_sum / grades_count, 1)
print('Average of entered grades:', average)
````

The following program generates random numbers and selects five even numbers from them and calculates their sum.

````python
import random

even_sum = 0
even_count = 0

while even_count < 5:
    number = random.randint(1, 10)
    if number % 2 == 0:
        print(number, 'even')
        even_count += 1
        even_sum += number
    else:
        print(number)

print('Sum of found even numbers: ', even_sum)
````

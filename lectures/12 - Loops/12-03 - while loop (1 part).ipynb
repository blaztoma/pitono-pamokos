# `while` Loop (Part 1)

## About the `while` Loop

It is most often used when we don't know how many times the loop should be executed. For example:

- we read a file, but we don't know how many lines of text are in it;
- we try to get data from the user, but we don't know when they will be able to provide correct data;
- we perform calculations and want to do them until we reach a suitable result;
- ...

## Syntax

```
while condition:
	# executed code if the condition is True
```

**condition** - the loop's condition statement; if it returns a `True` answer, then the loop executes an iteration.

The loop executes iterations as long as the condition is still True.

## Examples

Let's recall how we create a number sequence from 0 to 4 using the `range()` function and a `for` loop:

````python
for number in range(5):
    print(number)
````

We can do the same using a `while` loop:

````python
number = 0

while number < 5:
    print(number)
    number += 1
````

This is how a `while` loop looks when the range start and end are specified with variables:

````python
start = 10
end = 20

current = start

while current <= end:
    print(current)
    current += 1
````

This is how we output text several times using a `while` loop:

````python
text = 'We output this text'
remaining = 4

while remaining > 0:
    print(text)
    remaining -= 1
````

In this example, we output numbers until a generated random number is less than 10.

````python
from random import randint

number = 1

while number < 10:
    number = randint(1, 20)
    print(number)
````

In this example, we consider the case where we buy goods as long as there is inventory greater than zero.

````python
from random import randint

quantity = 20

while quantity > 0:
    print('Quantity before purchase:', quantity)
    purchased = randint(1, 10)
    print('Purchased:', purchased)
    quantity -= purchased
    print('Remaining:', quantity)
    print()
````

Let's recall how we can display information more conveniently using formatting. The example is the same.

````python
from random import randint

quantity = 20

while quantity > 0:
    purchased = randint(1, 10)
    print(f'Quantity before purchase: {quantity:<4} | Purchased: {purchased:<4} | Remaining: {quantity - purchased}')
    quantity -= purchased
````

Let's take the same example and allow the user to decide how many items to buy.

````python
item_quantity = 24

while item_quantity > 0:
    print(f'Available item quantity: {item_quantity}')
    purchased = int(input('How many do you want to buy? '))
    print(f'You purchased: {purchased} \n')
    item_quantity -= purchased
````

The previous example allowed buying items into the negative. Let's check if we can buy that much, otherwise we'll buy only as much as we have left.

````python
item_quantity = 24

while item_quantity > 0:
    print(f'Available item quantity: {item_quantity}')
    purchase_amount = int(input('How many do you want to buy? '))
    if purchase_amount > item_quantity:
        purchase_amount = item_quantity
    print(f'You purchased: {purchase_amount} \n')
    item_quantity -= purchase_amount
````

A `while True` loop is infinite. However, we can escape from it using the `break` operation. In this case, we escape when a random number is divisible by both 7 and 2.

````python
import random

while True:
    number = random.randint(1, 100)
    print(number)

    if number % 7 == 0 and number % 2 == 0:
        print(f'number {number} is divisible by 7 and 2')
        break
````

We can achieve the same result without using the `break` operation. In this case, we introduce a boolean variable `keep_searching`. When we change its value, we exit the loop.

````python
from random import randint

keep_searching = True

while keep_searching:
    number = randint(1, 100)
    print(number)

    if number % 7 == 0 and number % 2 == 0:
        print(f'number {number} is divisible by 7 and 2')
        keep_searching = False
````

We can move the division condition into the `while` loop description. This way we won't need to use a separate variable.

````python
import random

number = 1

while number % 7 != 0 or number % 2 != 0:
    number = random.randint(1, 100)
    print(number)

print(f'Number {number} is divisible by 7 and 2.')
````

Example of how we can check if a given number is prime. We organize a `for` loop and if we find a number by which the given number is divisible, we stop the loop.

````python
number = 57
is_prime = True

for divisor in range(2, number):
    if (number % divisor == 0):
        is_prime = False
        break

if is_prime:
    print(f'number {number} is prime')
else:
    print(f'number {number} is not prime')
````

The following program will search for the first random number that is prime. For this, we use a `while` loop and the code above.

````python
from random import randint

is_prime = False

while not is_prime:

    number = randint(1, 1000)
    is_prime = True

    for divisor in range(2, number):
        if (number % divisor == 0):
            is_prime = False
            break

    if is_prime:
        print(f'number {number} is prime')
    else:
        print(f'number {number} is not prime')
````

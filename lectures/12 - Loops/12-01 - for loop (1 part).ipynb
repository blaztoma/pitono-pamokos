# Loops

Loops are used to repeat the same actions.
There are two types of loops:
  - `for`
  - `while`

Each loop has certain characteristics, so in each case you need to decide which loop is more appropriate to use.

# `for` Loop (Part 1)

## About the `for` Loop

This loop is most often used when the number of times it needs to be executed is known, or when iterating over a dataset (list of numbers, set of elements, …).

## Syntax

```
for element in iterable_object:
	# executed code
```

- **iterable_object** - a dataset that needs to be processed.
- **element** - one element, one part, from `iterable_object`.

**iterable_object** can be:

```
[40, 32, 73]
"hello"
range(1, 10)
...
```

## Examples

Code that outputs 5 numbers without using a loop.

````python
print(0)
print(1)
print(2)
print(3)
print(4)
````

With a loop, we can do this more concisely.

````python
for i in range(5):
    print(i)
````

We can specify a specific interval for the loop (from 0 to 5).

````python
for i in range(0, 5):
    print(i)
````

We can change that interval as needed (from 1 to 6).

````python
for i in range(1, 6):
    print(i)
````

With this form of loop, the loop variable starts from 0, but inside the loop we can use this variable as needed (in this case, numbers starting from 1 will be output).

````python
for i in range(5):
    print(i+1)
````

With a loop, we can conveniently output not just numbers (which would be tedious without a loop).

Code that outputs the same string three times without a loop.

````python
print('hello')
print('hello')
print('hello')
````

Using a loop, the code becomes shorter.

````python
for _ in range(3):
    print('hello')
````

Another example with string formatting. Without a loop, we could format strings like this.

````python
print('line no. 1')
print('line no. 2')
print('line no. 3')
print('line no. 4')
````

Using a loop, we can do this more easily.

````python
for i in range(1, 5):
    print(f'line no. {i}')
````

A few more examples with numbers. Output numbers in the range [5, 10].

````python
for num in range(5, 11):
    print(f'number: {num}')
````

And... output numbers in the range [5, 15].

````python
for i in range(5, 16):
    print(f'number i = {i}')
````

Example of selecting even numbers in the range [2, 10].

````python
for number in range(2, 11):
    if number % 2 == 0:
        print(f'number {number} is even')
````

We can select even numbers more easily by specifying a step of 2.

````python
for number in range(2, 11, 2):
    print(f'number {number} is even')
````

A slightly more complex example, selecting numbers that are divisible by 2 or 5.

````python
for num in range(1, 13):
    if num % 2 == 0 or num % 5 == 0:
        print(f'number {num} is divisible by 2 or 5')
````

When specifying ranges, we can use not just numbers, but also variables.

````python
count = 5

for i in range(count):
    print(i)
````

More variables specifying the range bounds.

````python
start = 7
end = 10

for i in range(start, end + 1):
    print(f'number {i}')
````

Different way of declaring variables and using them in a loop.

````python
start, end = 7, 10

for i in range(start, end + 1):
    print(f'number {i}')
````

Using variables in a loop.

````python
name = input('Enter your name: ')
count = int(input('How many times do you want to repeat it: '))

for i in range(count):
    print(name)
````

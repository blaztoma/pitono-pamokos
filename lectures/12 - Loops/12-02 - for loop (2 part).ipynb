# `for` Loop (Part 2)

## Syntax

```
for element in iterable_object:
	# executed code
```

- **iterable_object** - a dataset that needs to be processed.
- **element** - one element, one part, from `iterable_object`.

**iterable_object** can be:

```
[40, 32, 73]
"hello"
range(1, 10)
...
```

## Examples

We will iterate through and print each character of the text.

````python
for char in "my text":
    print(char)
````

Let's recall how to print a character or string a certain number of times.

````python
symbol = 'a'
print(symbol * 10)
````

In this example, we will iterate through each character of the word and print it 10 times.

````python
text = 'apple'

for char in text:
    print(char * 10)
````

Just as we iterated through characters, we can also iterate through numbers.

````python
for element in [7, 4, 5, 32, 14, 78]:
    print(element)
````

In this example, we iterate through numbers and print which ones are even and which are odd.

````python
for number in [6, 7, 1, 5, 9, 7, 3]:
    if number % 2 == 0:
        print(f'Number {number} is even.')
    else:
        print(f'Number {number} is odd.')
````

When iterating through a dataset (characters, numbers, and other data), we can stop the loop using the keyword `break`.

````python
print("start")

for char in "oak":
    if char == "k":
        break
    print(char)

print("end")
````

Just as we can stop a loop, we can also continue it, skipping the following operations. For this we use the word `continue`.

````python
print("start")

for char in "oakland":
    if char == "k":
        continue
    print(char)

print("end")
````

Example where we iterate through numbers until the first negative one. Then we stop the loop using `break`. If the loop had not been stopped, then the `all_positive` variable would have remained `True`.

````python
all_positive = True

for number in [6, 1, -7, 5, -1, 4, 5]:
    if number < 0:
        all_positive = False
        break

print('All positive:', all_positive)
````

A small addition to the program above. We print whether all numbers are positive or if there is at least one negative number. For this we use an `if` ... `else` conditional statement.

````python
all_positive = True

for number in [6, 1, -7, 5, -1, 4, 5]:
    if number < 0:
        all_positive = False
        break

if all_positive:
    print('All numbers are positive.')
else:
    print('There is at least one negative number.')
````

Below you will see how a loop within a loop (nested loop)  is organized. The inner loop, whose loop variable is `j`, is executed three times. This is as many times as the outer loop, whose loop variable is `i`, is executed.

````python
for i in range(1, 4):

    print('Start of first loop, i =', i)

    for j in range(1, 4):
        print('Second loop, j =', j)

    print('End of first loop iteration.')
````

Here is shown a practical example of a loop within a loop. True, in a real case, the sets of students and their grades would be lists. In this case, for simplicity, we use `range()`.

````python
for student in range(1, 5):
    print(f'Student no. {student} grades: ', end='')
    for grade in range(1, 6):
        print(grade, end=' ')
    print()
````

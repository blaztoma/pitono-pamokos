# Using else

## About using else

- `else` can be translated as `otherwise` or `in other cases`.
- The `else` block will always be executed when no other condition was evaluated as true.
- In the `else` part, you don't need to write any condition. If you want to add a condition, then use the `elif` construct.
- The `else` part is written only once, at the end of the entire `if` construct.
- `else`, like `elif` blocks, is optional.
- We often use the `else` block when:
  - we want to ensure that the condition doesn't leave unhandled options;
  - printing an error message;
  - we want to allow repeating an action;
  - we want to detect an error.

## `else` syntax:

```
if condition:
	# this block's code is executed if the condition is met
else:
	# this block's code is executed if the condition is not met
```

else syntax when using the `elif` construct:

```
if condition_1:
	# this block's code is executed if condition_1 is met
elif condition_2:
  # this block's code is executed if condition_2 is met
# ...
elif condition_n:
  # this block's code is executed if condition_n is met
else:
	# this block's code is executed if none of the above conditions are met
```

## Examples

### Number comparison

````python
age = int(input('Enter age:'))

if age >= 18:
  print("Person is of legal age")
else:
  print("Person is a minor")
````

### Text comparison

````python
file = 'files/lectures/notes.txt'

if file.endswith('.txt'):
    print('Text file.')
    print('Starting work with file.')
else:
    print('Not a text file')
    print('Please select an appropriate file.')
````

### Number guessing

````python
from random import randint

random_number = randint(1, 10)

guess = int( input('Guess a number from 1 to 10: ') )

if random_number == guess:
    print('You guessed it!')
else:
    print('Unfortunately, you failed :(')
````

### Detecting an incorrect option

Besides conditional statements whose value is `True` or `False`, we can also get a `False` value if:
- we have an empty object;
- we have an empty string variable;
- None value is specified;
- the value is 0.

````python
file = 'files/lectures/notes.txt'

if file:
    print('Reading file...')
else:
    print('Please specify a file...')
````

````python
file = None

if file:
    print('Reading file...')
else:
    print('Please specify a file...')
````

````python
name = input('Enter your name:')

if name:
    print('Hello, ' + name + '!')
else:
    print('You entered nothing!')
````

## Condition check in one line (inline if / ternary operator)

We can shorten the if statement by using the ternary operator. This way we can write a multi-line `if` statement in one line.

Syntax:

```
[what to do in True case] if [check condition] else [what to do in False case]
```

If the grade is 10, then the evaluation is "Excellent", and if the grade is anything else, then the evaluation is "Could be better".

````python
grade = 10
evaluation = 'Excellent!' if grade == 10 else 'Could be better!'
print(evaluation)
````

````python
grade = 10
evaluation = 'Could be better!'

if grade == 10:
    evaluation = 'Excellent!'

print(evaluation)
````

If the customer has more than 100 points, they are a 'Gold' customer, otherwise they are a 'Silver' customer.

````python
points = 110
type = 'Gold' if points > 100 else 'Silver'

print(type)
````

````python
points = 110
type = 'Gold' if points > 100 else 'Silver'
print(points, type)

points = 50
print(points, type)

type = 'Gold' if points > 100 else 'Silver'
print(points, type)
````

sAs we can see - we saved several lines. However - ternary operators should be used carefully. Programmers still debate whether they are worth using. It is believed that they can reduce program readability and contribute to errors.

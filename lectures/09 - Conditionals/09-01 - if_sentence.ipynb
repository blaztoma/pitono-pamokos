# Program/control flow and if conditional statements

Computers and their programs would be unintelligent and hardly useful things if they couldn't make decisions. How intelligent they are is also a question worth discussing, but chatGPT, chess-playing programs and others have shown that computers can make decisions. And even - quite good ones. Here we will examine how a computer can make the simplest decisions that determine the program's execution flow. That is - which operations will be executed depending on the situation.

To steer the program's operation in a certain direction, we need to make a decision based on various conditions relevant to the program. That is, we need to compare what is relevant to the program, for example:

````python
password = input("Enter password: ")
if password == "admin":
  print("Welcome, you are logged in")
else:
  print("You failed to log in...")
````

We see that we compared the variable `password` with the text `admin`. Did you notice that the comparison uses two equal signs `==`? This is because in Python, one equal sign is reserved for the assignment statement. Of course - such a comparison is not enough, there are several other comparison operators, sometimes also called *comparators*. They are presented in the table.

Operator | Description | Example
---------|-------------|--------
`==` | is equal | `x == y`
`!=` | is not equal | `x != y`
`>` | is greater | `x > y`
`<` | is less | `x < y`
`>=` | is greater or equal | `x >= y`
`<=` | is less or equal | `x <= y`

As you can see, only the inequality sign is more interesting, which starts with an exclamation mark. It just so happens that in Python and various other languages, this sign means - NOT.

## If statement structure

An if conditional statement can consist of several parts:
* `if` - the mandatory part, after which actions are written that are executed when the condition is met.
* `elif` / `else if` - if you need to check more conditions, you can write as many else if parts as needed.
* `else` - written at the very end, once, if you need to cover all conditions that don't fall into other parts.

The simplest form of an `if` statement is:
```
if condition:
  actions...
```
That is, the keyword `if` is written, the condition is specified and a colon is written at the end. After that, actions are specified that are indented from the edge.

The condition's actions will be executed only if we get a true answer (`True` type) in the condition.

### Checking arithmetic operations

````python
number = 8

if 5 + 3 >= 6:
    print('5 + 3 >= 6 check worked')

if number + 2 > 10:
    print(number, '+ 2 > 10 check worked')

if number * 2 > 0:
    print(number, '* 2 > 0 check worked')

if 8 * 2 < 8 + 6:
    print('8 * 2 < 8 + 6 check worked')
````

### Is the number divisible?

````python
number = int( input('Enter a number: ') )

print(f'{number}:')

if number % 2 == 0:
    print('divisible by 2')
if number % 3 == 0:
    print('divisible by 3')
if number % 4 == 0:
    print('divisible by 4')
if number % 5 == 0:
    print('divisible by 5')
if number % 6 == 0:
    print('divisible by 6')
if number % 7 == 0:
    print('divisible by 7')
if number % 8 == 0:
    print('divisible by 8')
if number % 9 == 0:
    print('divisible by 9')
````

As you can see - we use the remainder operation. If the remainder equals zero, it means the number is divisible.

### Is the number not divisible?

````python
number = 8

if number % 2 != 0:
    print('Number is not divisible by 2')

if number % 3 != 0:
    print('Number is not divisible by 3')

if number % 4 != 0:
    print('Number is not divisible by 4')
````

A number is not divisible if the remainder is *not equal* to zero.

## Boolean values

An if statement's condition can be very complex, but when the condition is evaluated, one value is obtained that tells whether the condition was true or false. That value is a boolean value, which can be `True` or `False` (true/false). If we try to give the `print` function a condition, Python will print the result.

````python
number = 14

print(number > 0)
print(number >= 10)
print(number < 0)
print(number <= 10)
print(number == 10)
print(number != 10)
````

Let's look more closely at what `True` and `False` values mean to Python. Try to explain the printed results.

````python
print(1 == 1)
print(True == 1)
````

````python
print(True == False)
print(True == 0)
````

````python
print(False == 0)
````

If a boolean value is calculated earlier, we can compare it or let Python decide by default. The second variant is more commonly used.

````python
is_driving = True

if is_driving == True:
    print('First if says they are driving.')

if is_driving:
    print('Second if says they are driving.')
````

## if's... inside if's

Often there is a case when one condition is not enough. Then we can use `if` inside another `if` block.

````python
crosswalk = True
cars_moving = False

if crosswalk:
  if cars_moving == False:
    print("You can cross the crosswalk")
````

True, this statement can be written shorter using logical operators. We will examine them later. You can think about using an `if` block inside another `if` when various conditions in one `if` statement become hard to read and understand.

# Using elif / else if

## elif / else if features

- When using only the if part, you probably noticed that you check only one condition. `elif` / `else if` allows you to check multiple conditions (one after another). For example:
  - Is the number positive? If not - further check: maybe the number is negative?
  - Is the user 18 years old? If not - further check: maybe the user is 16 years old? If not - next check - maybe the user is less than 16?
- You can use as many `elif` blocks as needed.
- Each `elif` part is responsible for processing a new condition.
- If a true condition is found when evaluating different conditions, then the code in that block is executed and further conditions are not checked. That is, only the first true condition found is executed.

## elif / else if syntax

```
if condition_1:
	# this block's code is executed if condition_1 value is True
elif condition_2:
	# this block's code is executed if condition_2 value is True
elif condition_3:
	# this block's code is executed if condition_3 value is True
```

## Examples

### Number comparison

````python
if 5 > 8:
    print('5 > 8 worked')
elif 7 > 3:
    print('7 > 3 worked')
elif 1 == 1:
    print('1 == 1 worked')
````

### Comparing two numbers

````python
number_1 = int(input("Enter number 1: "))
number_2 = int(input("Enter number 2: "))

if number_1 > number_2:
  print("Number 1 is greater than number 2.")
elif number_1 < number_2:
  print("Number 1 is less than number 2.")
elif number_1 == number_2:
  print("Number 1 equals number 2.")
````

### Menu selection

````python
print("1. Play")
print("2. Leaderboard")
print("3. Settings")
print("4. Exit")

choice = int( input('Enter menu (1-4): ') )

if choice == 1:
    print('Get ready to play hard!')
elif choice == 2:
    print('We will present the leaderboard next...')
elif choice == 3:
    print('Set up graphics and sound settings...')
elif choice == 4:
    print('Program is shutting down...')
````

### Finding a range

````python
points = 40

if points >= 50:
    print('You are a genius!')
elif points >= 30:
    print('You are an excellent player')
elif points >= 10:
    print('You are a good player')
elif points < 5:
    print('You are a beginner player')
````

### VIP entry or ticket?

````python
money = 10
vip = True

if vip: # can also be vip == True
    print('You have VIP and can enter.')
elif money > 5:
    print('You can buy a ticket')
````

### Text comparisons

````python
role = 'moderator'

if role == 'admin':
    print('You have been granted all system capabilities.')
elif role == 'moderator':
    print('Your rights are limited to comment moderation.')
elif role == 'user':
    print('You cannot manage the website.')
````

````python
name = 'John'

if name == 'john':
    print('john in lowercase.')
elif name == 'John':
    print('John with capital letter.')
elif name == 'JOHN':
    print('JOHN, all capital letters.')
````

````python
name = 'Peter'

if 'ABC' in name:
    print('Name', name, 'contains ABC letters.')
elif name.startswith('Pe'):
    print('Name', name, 'starts with Pe...')
elif name.endswith('er'):
    print('Name', name, 'ends with -er.')
````

````python
search_phrase = 'SeCret'

if search_phrase == 'secret':
    print('Found, case sensitive.')
elif search_phrase.lower() == 'secret':
    print('Found, case insensitive.')
elif search_phrase.lower() != 'secret':
    print('Not found.')
````

# Output formatting

Often we need to calculate various values, combine them, and print them nicely. Let's look at a few ways to do this.

If we use the `print()` function and separate variables with a comma, then Python itself will create a string for output:

````python
brand = 'Citroen'
model = 'Xsara'
year = 2002 
mileage = 148974

print(brand, model, 'was manufactured in', year, 'and has driven', mileage, 'km.')
````

Strings can be joined before printing. This is one of the most commonly used ways. However, when concatenating strings, you need to convert variables of other types to the string type. For that, use the `str()` function.

````python
brand = 'Citroen'
model = 'Xsara'
year = 2002
mileage = 148974

# This code will not work (it will show an error), because Python can't concatenate a string and a number.
# You can uncomment this line and see for yourself.  
# sentence = brand + " " + model + " was manufactured in " + year + " and has driven " + mileage + " km."

# But this code is good, because it converts numbers to the string type
sentence = brand + " " + model + " was manufactured in " + str(year) + " and has driven " + str(mileage) + " km."

print(sentence)
````

If we want to avoid type conversion, we can form a text string using the `%` formatting operation. This method isn't recommended by Python itself, but you can see what it looks like:

````python
brand = 'Citroen'
model = 'Xsara'
year = 2002
mileage = 148974

print("%s %s was manufactured in %s and has driven %s km." % (brand, model, year, mileage))
````

A slightly better way is to use `format`:

````python
brand = 'Citroen'
model = 'Xsara'
year = 2002
mileage = 148974

print("{1} {0} was manufactured in {2} and has driven {3} km.".format(model, brand, year, mileage))
````

In newer Python (from version 3.6), an even more convenient and better way to format output text appeared: f-strings (`f""` or `f''`). These f-strings are more resistant to errors and faster than the `%` or `format` methods. Use them.


````python
brand = 'Citroen'
model = 'Xsara'
year = 2002
mileage = 148974

print(f'{brand} {model} was manufactured in {year} and has driven {mileage} km.')
````

And finally, if you need to format multi-line text, do it like this:

````python
brand = 'Citroen'
model = 'Xsara'
year = 2002
mileage = 148974

sentence = f"""
  {brand} {model} 
  was manufactured in {year}
  and has driven {mileage} km.'
"""

print(sentence)
````

# Console input

One way to input data into a program is to ask the user to enter the required information in the console. For this, the `input()` function is used.

When the request to enter information is displayed separately:

````python
print('Enter a name:')
name = input()
print(name)
````

When the request to enter information is displayed together:

````python
name = input('Enter a name:')
print(name)
````

What type of information do we get from `input`?

````python
print('Enter any number:')
number = input()

print('Entered:', number, type(number))
````

We get a number from the previous input:

````python
number = int(number)
print('Converted:', number, type(number))
````

We can also implement number input in one line:

````python
number = int(input('Enter an integer:'))
print(number)
````

When we want to get an entered floating-point number:

````python
number = float(input('Enter a number:'))
print(number)
````

To remind about f-strings, let's enter two numbers on separate lines and add them:

````python
number_1 = input('Enter the first number')
number_2 = input('Enter the second number')
print(f'{number_1} + {number_2} = {number_1 + number_2}')
````

Note that in the previous example, the received input is of type `str`, so the addition (`+`) performed text concatenation, not mathematical addition. This could be fixed like this:

````python
number_1 = int(input('Enter the first number'))
number_2 = int(input('Enter the second number'))
print(f'{number_1} + {number_2} = {number_1 + number_2}')
````

Or like this:

````python
number_1 = input('Enter the first number')
number_2 = input('Enter the second number')
print(f'{number_1} + {number_2} = {int(number_1) + int(number_2)}')
````

This time we'll enter those two numbers on one line:

````python
print('Enter two numbers, separating them with a space:')
number_1, number_2 = input().split()
print(f'{number_1} + {number_2} = {int(number_1) + int(number_2)}')
````

````python
print('Enter two numbers separated by a space:')
number_1, number_2 = input().split()
number_1 = int(number_1)
number_2 = int(number_2)
print(f'{number_1} + {number_2} = {number_1 + number_2}')
````

````python
print('Enter three numbers separated by a space:')
number_1, number_2, number_3 = [ int(number) for number in input().split() ]
print(f'{number_1} + {number_2} + {number_3} = {number_1 + number_2 + number_3}')
````

It's also possible like this:

````python
print('Enter two numbers (separated by pressing enter):')
first = int(input())
second = int(input())

total = first + second
print('sum of numbers:', total)
````

More input:

````python
name = input('Enter name: ')
age = int( input('Enter age: ') )

print('Entered information:')
print(name, type(name))
print(age, type(age))
````

# Lists (`lists`) (Part 4)

## List slicing

If we want to cut out certain parts of a list, we can use a technique called `list slicing`.

Slicing is done by specifying parameters inside `[ ]` brackets.

After slicing, a new list is returned.

## Syntax

When slicing a list, we can specify the desired ranges like this:

```
list_name[start:end]
list_name[start:end:step]
```

Explanations:

- `list_name` - the list being processed.
- `start` - the index where slicing starts (included).
- `end` - the index where slicing ends (not included).
- `step` - how many elements to skip after each included element.

When slicing, it is not necessary to specify all values.

Everything is taken from the start up to the specified `end`:

```
list_name[:end]
```

Everything is taken from the specified `start` to the end of the list:

```
list_name[start:]
```

Everything is taken from start to end:

```
list_name[:]
```

Everything is taken from start to the specified `end`, using the specified `step`:

```
list_name[:end:step]
```

Everything is taken from the specified `start` to the end, using the specified `step`:

```
list_name[start::step]
```

And so on.

## Examples

A few examples using the syntax described above, where start and end are specified.

````python
words = ['tree', 'pear', 'paper', 'tree', 'amber']

print('All:', words)

print('When [0]:', words[0])      # [index]
print('When [0:2]:', words[0:2]) # [start:end]
print('When [1:4]:', words[1:4]) # [start:end]
````

A few examples using the syntax described above, where start or end is not specified.

````python
words = ['tree', 'pear', 'paper', 'tree', 'amber']

print('All:', words)

print('When [2:]:', words[2:])  # [start:end]
print('When [:3]:', words[:3]) # [start:end]
````

A few examples using `start:end:step`.

````python
words = ['tree', 'pear', 'paper', 'tree', 'amber']

print('All:', words)

print('When [0::1]:', words[0::1]) # [start:end:step]
print('When [::1]:', words[::1])   # [start:end:step]
print('When [0::2]:', words[0::2]) # [start:end:step]
print('When [::2]:', words[::2])   # [start:end:step]
````

````python
numbers = [7, 8, 6, 5, 8, 7, 6, 5, 2, 3]

print('All:', numbers)

print('When [2:8:2]:', numbers[2:8:2]) # [start:end:step]
print('When [4::3]:', numbers[4::3])  # [start:end:step]
````

````python
cars = ['audi', 'bmw', 'volvo', 'mercedes', 'opel']

selected = cars[2:] # [start:end]

print(cars)
print(selected)
````

A cool trick! If we set the step to `-1`, we get a reversed list.

````python
words = ['paper', 'computer', 'book', 'paper', 'tree']

# reverses
words.reverse()

print(words)

# also reverses, and creates a new list
print(words[::-1]) # [start:end:step]
````

An example where we make a copy of a list. In this case, if we change a value in the copy, the original list will not change.

````python
cars = ['audi', 'bmw', 'volvo', 'mercedes', 'opel']
cars[4] = 'bentley'

print('CARS:', cars)

copy_list = cars[:] # [start:end]
print('COPY:', copy_list)

cars[1] = 'NEW'
print('CARS:', cars)
print('COPY:', copy_list)
````

An example where we assign a list to another variable. In this case, if we change a value through the new variable, the list referenced by the original variable will also change. Both variables use the same list.

````python
cars = ['audi', 'bmw', 'volvo', 'mercedes', 'opel']
cars[4] = 'bentley'

print('CARS:', cars)

copy_list = cars
print('COPY:', copy_list)

cars[1] = 'NEW'
print('CARS:', cars)
print('COPY:', copy_list)
````

This is how we can check whether list values match and whether variables use the same list in memory.

````python
cars = ['audi', 'bmw', 'volvo', 'mercedes', 'opel']
copy_list = cars[:] # [start:end]

print('Do list values match:', cars == copy_list)
print('Do the lists use the same memory:', cars is copy_list)
````

An example where variables use the same list in memory.

````python
cars = ['audi', 'bmw', 'volvo', 'mercedes', 'opel']
copy_list = cars

print('Do list values match:', cars == copy_list)
print('Are the lists the same object in memory:', cars is copy_list)
````

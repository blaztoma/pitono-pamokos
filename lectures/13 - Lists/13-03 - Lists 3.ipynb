# Lists (`lists`) (Part 3)

## Examples

Sorting uses the list object's `sort()` method. This method sorts the elements of the list.

```python
words = ['paper', 'computer', 'book', 'paper', 'tree']

# sorts the data in the list, modifying the original list
words.sort()

print(words)
```

For sorting, we can also use the `sorted()` function. This function creates a new list with sorted elements, but it does not modify the original list.

```python
words = ['paper', 'computer', 'book', 'paper', 'tree']

# sorts the data, but returns a new list
print(sorted(words))
```

This example illustrates exactly that.

```python
words = ['paper', 'computer', 'book', 'paper', 'tree']
print(sorted(words)) # sorted returns a new list
print(words)
```

The list method `reverse()` reverses the elements.

```python
words = ['paper', 'computer', 'book', 'paper', 'tree']

# reverses the order
words.reverse()

print(words)
```

By combining `reverse()` with `sort()`, we can get a list sorted in descending order.

```python
words = ['paper', 'computer', 'book', 'paper', 'tree']

# sorts
words.sort()

# reverses the order
words.reverse()

print(words)
```

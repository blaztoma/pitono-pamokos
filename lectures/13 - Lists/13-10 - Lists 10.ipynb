## Examples

```python
matrix = [
    [8, 7, 5, 3],
    [4, 7, 5, 2],
    [4, 7, 8, 3],
    [5, 2, 3, 1]
]

print('Entire matrix:', matrix)
print('Second row:', matrix[1])
print('Third element of the second row:', matrix[1][2])
```

```python
student_grades = [
    ['Jonas', 7, 8, 5, 6, 3],
    ['Petras', 8, 9, 8, 8, 9, 10],
    ['Ona', 7, 8],
    ['Birutė', 9, 9, 10, 10, 7, 8],
    ['Ugnius', 7, 7, 8, 9, 7]
]

for student in student_grades:
    print(student)
```

```python
student_grades = [
    ['Jonas', 7, 8, 5, 6, 3],
    ['Petras', 8, 9, 8, 8, 9, 10],
    ['Ona', 7, 8],
    ['Birutė', 9, 9, 10, 10, 7, 8],
    ['Ugnius', 7, 7, 8, 9, 7]
]

for student in student_grades:
    for el in student:
        print(el, end=' ')
    print()
```

```python
student_grades = [
    ['Jonas', 7, 8, 5, 6, 3],
    ['Petras', 8, 9, 8, 8, 9, 10],
    ['Ona', 7, 8],
    ['Birutė', 9, 9, 10, 10, 7, 8],
    ['Ugnius', 7, 7, 8, 9, 7]
]

for student in student_grades:
    print('student', student[0], 'grades: ', end='')
    for i in range(1, len(student)):
        print(student[i], end=' ')
    print()
```

```python
student_grades = [
    ['Jonas', [7, 8, 5, 6, 3]],
    ['Petras', [8, 9, 8, 8, 9, 10]],
    ['Ona', [7, 8]],
    ['Birutė', [9, 9, 10, 10, 7, 8]],
    ['Ugnius', [7, 7, 8, 9, 7]]
]

for student in student_grades:
    print('Student:', student[0])
    print('His/her grades:', student[1])
    print('His/her grades:', ', '.join( map(str, student[1]) ))
    print()
```

```python
student_grades = [
    ['Jonas', [7, 8, 5, 6, 3]],
    ['Petras', [8, 9, 8, 8, 9, 10]],
    ['Ona', [7, 8]],
    ['Birutė', [9, 9, 10, 10, 7, 8]],
    ['Ugnius', [7, 7, 8, 9, 7]]
]

print('All grades of all students:')

for student in student_grades:
    for grade in student[1]:
        print(grade, end=' ')
    print()
```

```python
matrix = [
    [8, 7, 5, 3],
    [4, 7, 5, 2],
    [4, 7, 8, 3],
    [5, 2, 3, 1]
]

sums = [ sum(row) for row in matrix ]
print('Row sums:', sums)

total_sum = sum(sums)
print('Total sum of the whole matrix:', total_sum)
```

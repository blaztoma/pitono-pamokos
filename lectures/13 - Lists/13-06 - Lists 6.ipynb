# Lists (`lists`) (Part 6)

## List unpacking (`list unpacking`)

## Examples

Let’s say we have a list like this:

```python
person_info = ['Jonas', 'Jonaitis', 23, 'KTU', 7]
print(person_info)
```

We can **unpack** it into separate variables.

```python
first_name, last_name, age, school, average = ['Jonas', 'Jonaitis', 23, 'KTU', 7]

print(first_name)
print(last_name)
print(age)
print(school)
print(average)
```

Here is how we unpack a student’s first name, last name, and their list of grades. Notice that an asterisk (`*`) is used to capture the list part.

```python
first_name, last_name, *grades = ['Petras', 'Petrauskas', 7, 8, 9, 6, 5]

print(first_name)
print(last_name)
print(grades)
```

Python will also understand this kind of assignment.

```python
first_name, last_name, *grades, school = ['Ona', 'Onutė', 7, 8, 9, 7, 5, 9, 'KTU']

print(first_name)
print(last_name)
print(grades)
print(school)
```

We can also unpack a string into separate variables. For that, we first convert it into a list.

```python
sentence = 'Jonas;Jonauskas;KTU;7;8;9'
info = sentence.split(';')
print(info)
```

After converting to a list, we can unpack its values.

```python
sentence = 'Jonas;Jonauskas;KTU;7;8;9'
first_name, last_name, school, *grades = sentence.split(';')

print(first_name)
print(last_name)
print(school)
print(grades)
```

However, the grades are still strings. So we need to convert them into a list of ints.

```python
sentence = 'Jonas;Jonauskas;23;KTU;7;8;9'
first_name, last_name, years, school, *grades = sentence.split(';')
years = int(years)
grades = list(map(int, grades))

print(first_name)
print(last_name)
print(years)
print(school)
print(grades)
```

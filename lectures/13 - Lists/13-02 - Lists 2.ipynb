# Lists (`lists`) (Part 2)

## Examples

Adding an element to the end and at a specified index.

```python
students = ['Justė', 'Tomas', 'Giedrius', 'Evelina']

# adds to the end
students.append('Jonas')

print(students)

# adds to a specified position
students.insert(1, 'Ugnė')

print(students)
```

Removing an element from the end of a list.

```python
numbers = [7, 5, 8, 9, 6]

print(numbers)

# removes the very last item
numbers.pop()

print(numbers)
```

You can remove more than one element from the end of a list.

```python
numbers = [7, 5, 8, 9, 6]
print(numbers)

numbers.pop()
numbers.pop()
print(numbers)
```

Using the `pop` method, you can remove an element from a specific position.

```python
numbers = [7, 5, 8, 9, 6]

print(numbers)

# removes from the specified position (index)
numbers.pop(0)

print(numbers)
```

The `clear()` method clears the list.

```python
numbers = [7, 5, 8, 9, 6]

print(numbers)

# completely clears the entire list
numbers.clear()

print(numbers)
```

The `index` method finds the first specified element and returns its index.

```python
numbers = [7, 5, 8, 9, 6]

# tells at which position in the list (which index) the specified value is
print(numbers.index(9))
```

Just like the `index` method finds numbers, it can also find strings.

```python
text = ['my', 'shoes', 'were', 'two']

# tells at which position in the list (which index) the specified value is
print(text.index('were'))
```

For searching, we can specify start and end indexes. This time the program will not find `batai`, because it will start searching from the third element.

```python
text = ['my', 'shoes', 'were', 'two']

# tells at which position in the list (which index) the specified value is,
# but starts searching at index 2 and stops at index 3
print(text.index('shoes', 2, 3))
```

This time the program will find `batai`, because it will start searching from the second element.

```python
text = ['my', 'shoes', 'were', 'two']

# tells at which position in the list (which index) the specified value is,
# but starts searching at index 1 and stops at index 3
print(text.index('shoes', 1, 3))
```

This is how we can find out whether a certain value exists in a specified list.

```python
words = ['paper', 'computer', 'book', 'paper', 'tree']

# tells whether such a value exists in the list
print('book' in words)
```

This time the program will not find the value `x` in the list of words.

```python
words = ['paper', 'computer', 'book', 'paper', 'tree']

# tells whether such a value exists in the list
print('x' in words)
```

This is how we can check whether a letter (character) exists in a string.

```python
# tells whether such a value exists in the text
print('m' in 'hi, my name is Ieva')
```

Another example with a list of cities.

```python
cities = ['Vilnius', 'Kaunas', 'Klaipėda', 'Šiauliai']

if 'Kaunas' in cities:
    print('Kaunas was found among the cities.')
else:
    print('Kaunas was not found among the cities.')

if 'Biržai' in cities:
    print('Biržai was found among the cities.')
else:
    print('Biržai was not found among the cities.')
```

Using the `count()` method, we can count how many times an element appears in a list.

```python
words = ['paper', 'computer', 'book', 'paper', 'tree']

# tells how many times this value is found in the list
print(words.count('popierius'))
```

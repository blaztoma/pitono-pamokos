````python
import random

count = int(input('How many numbers to generate? '))
numbers = []

for _ in range(count):
    numbers.append(random.randint(1, 100))

print(numbers)
````

````python
numbers = [7, 4, 7, 8, 3, 6, 8]

i = 0

while i < len(numbers):
    print('Index:', i)
    print('Number:', numbers[i])
    print()
    i += 1
````

````python
numbers = [78, 47, 24, 36, 24, 14, 25, 98, 73]

for n in numbers:
    if n % 2 == 0:
        print(f'number {n} is even')
    else:
        print(f'number {n} is odd')
````

````python
numbers = [78, 47, 24, 36, 24, 14, 25, 98, 73]

for n in numbers:
    output = f'number {n} is '
    if n % 2 == 0:
        output += 'even'
    else:
        output += 'odd'
    print(output)
````

````python
numbers = [8, 5, 7, 4, 2, 3]

even_sum = 0
odd_sum = 0

for n in numbers:
    if n % 2 == 0:
        even_sum += n
    else:
        odd_sum += n

print('numbers:', numbers)
print('sum of evens:', even_sum)
print('sum of odds:', odd_sum)
````

````python
numbers = [8, 5, 7, 4, 2, 3]

even_sum = 0
even_count = 0

for n in numbers:
    if n % 2 == 0:
        even_sum += n
        even_count += 1

average = round(even_sum / even_count, 2)

print('numbers:', numbers)
print('sum of evens:', even_sum)
print('count of evens:', even_count)
print('average of evens:', average)
````

````python
grades = [7, 4, 8, 7, 9, 6, 2, 5]

print('passing grades: ', end='')

for g in grades:
    if g >= 5:
        print(g, end=' ')

print()
````

````python
grades = [7, 4, 8, 7, 9, 6, 2, 5]

passing = []
failing = []

for g in grades:
    if g >= 5:
        passing.append(g)
    else:
        failing.append(g)

print('grades:', grades)
print('passing grades:', passing)
print('failing grades:', failing)
````

````python
first_student_grades = [8, 7, 9, 8, 7, 9, 10, 9]
second_student_grades = [9, 9, 10, 8, 7, 6, 9]

first_student_average = round(sum(first_student_grades) / len(first_student_grades), 2)
second_student_average = round(sum(second_student_grades) / len(second_student_grades), 2)

print('FIRST STUDENT')
print('grades:', first_student_grades)
print('average:', first_student_average)
print('SECOND STUDENT')
print('grades:', second_student_grades)
print('average:', second_student_average)
print()

if first_student_average > second_student_average:
    print("the first student's average is higher")
    difference = round(first_student_average - second_student_average, 2)
    print('difference between averages:', difference)
elif second_student_average > first_student_average:
    print("the second student's average is higher")
    difference = round(second_student_average - first_student_average, 2)
    print('difference between averages:', difference)
else:
    print('both students have the same average')
````

````python
grades = [7, 4, 8, 7, 9, 6, 2, 5]

failing_count = 0

for g in grades:
    if g < 5:
        failing_count += 1

print('grades:', grades)
print('number of failing grades:', failing_count)
````

````python
numbers = [8, 7, 9, 6, 5, 8, 4]

maximum = numbers[0]

for n in numbers:
    if n > maximum:
        maximum = n

print('largest number:', maximum)
````

````python
numbers = [8, 7, 9, 6, 5, 8, 4]

maximum = numbers[0]
max_index = 0

for i, n in enumerate(numbers):
    if n > maximum:
        maximum = n
        max_index = i

print('largest number:', maximum)
print('index of the largest:', max_index)
````

````python
import random

# empty list
numbers = []

# generating numbers
count = int(input("how many numbers do you want: "))
for _ in range(count):
    numbers.append(random.randint(1, 10))

# printing numbers
print('generated numbers:', numbers)

# sum of numbers
total = sum(numbers)
print('sum:', total)

# average of numbers
average = total / len(numbers)
print('average:', round(average, 2))
````

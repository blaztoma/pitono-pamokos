# Lists (`lists`) (Part 7)

## Examples

Printing all information manually:

```python
cars = ['audi', 'bmw', 'volvo', 'mercedes']

print(cars[0])
print(cars[1])
print(cars[2])
print(cars[3])
```

Printing all information using a loop:

```python
cars = ['audi', 'bmw', 'volvo', 'mercedes']

for car in cars:
    print(car)
```

```python
cities = ['Kaunas', 'Vilnius', 'Klaipėda', 'Šiauliai']

for city in cities:
    print(city)
```

If you need indexes:

```python
cities = ['Kaunas', 'Vilnius', 'Klaipėda', 'Šiauliai']

for index, city in enumerate(cities, start=0): # default is 0
    print(index, city)
```

```python
cities = ['Kaunas', 'Vilnius', 'Klaipėda', 'Šiauliai']

for index, city in enumerate(cities, start=1): # default is 0
    print(index, city)
```

```python
cities = ['Kaunas', 'Vilnius', 'Klaipėda', 'Šiauliai']

for index in range(len(cities)):
    print(index, cities[index])
```

```python
cities = ['Kaunas', 'Vilnius', 'Klaipėda', 'Šiauliai']

for index in range(len(cities)):
    print(index + 1, cities[index])
```

```python
cities = ['Kaunas', 'Vilnius', 'Klaipėda', 'Šiauliai']

for index in range(len(cities)):
    print(f'{index + 1}th city is {cities[index]}')
```

```python
cities = ['Kaunas', 'Vilnius', 'Klaipėda', 'Šiauliai']

index = 0

while index < len(cities):
    print(f'{index + 1}th city: {cities[index]}')
    index += 1
```

```python
people = ['Petras', 'Jonas', 'Ona', 'Birutė']

i = 0

while i < len(people):
    print(f'{i}: {people[i]}')
    i += 1
```

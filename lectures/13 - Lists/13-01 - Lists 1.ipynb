# Lists (`lists`) (Part 1)

## About lists

You can think of lists as a collection of data. Some possible examples:

- students' grades;
- words;
- products;
- students;
- cars;
- ...

## Creating a list

An empty list (no data):

```
list_name = []
```

For example:

````python
information = []
````


A list with data:

```
list_name = [value, value, value, ...]
```

For example:

````python
numbers = [7, 4, 5, 8, 3, 1]
````

````python
people = ['Jonas', 'Petras', 'Ona', 'Birutė']
````

## Displaying the whole list

If we want to display all the information in a list, we can do it like this:

````python
numbers = [7, 5, 6, 9, 7, 5]
print(numbers)
````

````python
people = ['Ona', 'Petras', 'Birutė', 'Jonas']
print(people)
````

## Adding to a list

If you want to add to an existing list, you can use the `append()` or `extend()` methods.

For example:

````python
numbers = [1, 2, 3, 4]

numbers.append(5)
numbers.append(6)
print(numbers)

numbers.extend([7, 8, 9])
print(numbers)
````

## Number of elements in a list

To find out how many elements a list contains, we can use the `len()` function:

```
len(list_name)
```

For example:

````python
numbers = [5, 2, 3, 5, 4]
print( len(numbers) )
````

## Indexing

If a list contains data, then each element in the list has its own index. Using that index, you can access or change the information.

Indexes are counted starting from 0.

Suppose we have this list:

```
grades = [7, 8, 5, 9, 10, 6]
```

Each list element will get the following indexes:

| List element | 7 | 8 | 5 | 9 | 10 | 6 |
|-|-|-|-|-|-|-|
| Assigned index | 0 | 1 | 2 | 3 | 4 | 5 |

If we have this list:

```
trees = ['Gluosnis', 'Ąžuolas', 'Liepa', 'Berželis', 'Obelis']
```

Each list element will get the following indexes:

| List element | 'Gluosnis' | 'Ąžuolas' | 'Liepa' | 'Berželis' | 'Obelis' |
|-|-|-|-|-|-|
| Assigned index | 0 | 1 | 2 | 3 | 4 |

Python also allows indexing from the end. In that case, numbering starts from -1, for example:

| List element | 'Gluosnis' | 'Ąžuolas' | 'Liepa' | 'Berželis' | 'Obelis' |
|-|-|-|-|-|-|
| Assigned index (from the end) | -5 | -4 | -3 | -2 | -1 |

### Index usage syntax

To get information:

```
list_name[index]
```

To update information:

```
list_name[index] = new_value
```

## Examples

Let's look at how different types of lists are represented.

````python
numbers_list = [1, 2, 3, 4, 5, 6]
letters_list = ['a', 'b', 'c']
mixed_list = [1, 2, 'a', True]

print(numbers_list)
print(letters_list)
print(mixed_list)
````

Look at the program and name the printed values in your head. Run the program and check.

````python
numbers = [78, 65, 95, 21, 23, 14, 74, 147]

print(numbers[0])
print(numbers[2])
print(numbers[4])
print(numbers[5])
````

Various index manipulations are shown here.

````python
numbers = [47, 54, 25, 36, 87]

print('List:', numbers)
print('First number in the list:', numbers[0])
print('Number of elements in the list:', len(numbers))
print('Last number in the list:', numbers[-1])
print('Last number in the list:', numbers[len(numbers) - 1])
````

See how we can add items to lists.

````python
numbers = []

print(numbers)

# add the specified value to the list (appended to the end)
numbers.append(87)
numbers.append(95)

print(numbers)

# add several specified values to the list (appended to the end)
numbers.extend([147, 258, 399])

print(numbers)
````

A simple calculation of the average of list elements. Think about how you could do this using a loop.

````python
grades = [8, 7, 9, 9, 8]

print("Student's grades:", grades)

total = grades[0] + grades[1] + grades[2] + grades[3] + grades[4]
average = total / len(grades)

print('Calculated sum:', total)
print('Grade average:', average)
````

An example of replacing list elements.

````python
names = ['Agnė', 'Tomas', 'Gintarė', 'Petras']

print(names)

names[0] = 'Paulina'
names[3] = 'Justas'

print(names)
````

In the examples, the same operations can be performed not only with numbers, but also with strings and other items.

````python
files = [
    'naujienos/pirma.txt',
    'naujienos/antra.txt',
    'naujienos/trecia.txt',
    'naujienos/ketvirta.txt',
    'naujienos/penkta.txt'
]

print('News files:', files)
print('Number of news items:', len(files))
print()

another_one = 'naujienos/sesta.txt'
files.append(another_one)

print('Additionally added news item:', another_one)
print('News files:', files)
print('Number of news items:', len(files))
````

When we use `range()`, we get a `range` object. To get a sequence of numbers, we need to convert the `range` object into a list.

````python
print(range(1, 100))
print(list(range(1, 100)))
````

Here is an example of building a list from random numbers.

````python
from random import randint

numbers = []
print(numbers)

new_num = randint(1, 10)
new_num2 = randint(1, 15)

numbers.append(new_num)
numbers.append(new_num2)

print(numbers)
````

Extending a list with a few random values.

````python
# This line will work if you have executed the previous cell.
# from random import randint

numbers = []
print(numbers)

numbers.extend([ randint(1, 10), randint(1, 10), randint(1, 10) ])
print(numbers)
````

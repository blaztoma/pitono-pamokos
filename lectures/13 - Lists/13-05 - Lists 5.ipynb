# Lists (`lists`) (Part 5)

## Splitting and joining lists

Lists can be joined into strings, and strings can be split into lists.

- When we want `list` -> `str`, we use the `join` method.
- When we want `str` -> `list`, we use the `split` method.

## `join`

`join` is used when we want to join a list into a string.

Syntax:

```
'<separator between each element>'.join(<list>)
```

## `split`

`split` is used when we want to split a string into a list.

Syntax:

```
<string>.split(<separator to split by>)
```

## Examples

We turn the list `list_items` into a sentence by joining its elements with spaces.

````python
list_items = ['hi', 'my', 'name', 'is', 'JOJO']
sentence = ' '.join(list_items)
print(sentence)
````

This is the same as...

````python
sentence = ' '.join(['hi', 'my', 'name', 'is', 'JOJO'])
print(sentence)
````

We can turn a list into a string using various separators (or no separator).

````python
cities = ['Kaunas', 'Vilnius', 'Klaipėda', 'Jonava', 'Biržai']

print(cities)

print(''.join(cities))
print(' '.join(cities))
print('|'.join(cities))
print(' / '.join(cities))
````

We can convert not only lists of strings into strings, but also lists of numbers. However, it is not that simple. Try it ;)

````python
numbers = [7, 4, 8, 5, 3]
print(' '.join(numbers))
````

To convert numbers to a string, we need to convert each element to a string. This example shows two ways to do that.

````python
grades = [7, 8, 7, 9, 10, 8]
print(grades)

grades_sentence1 = ', '.join(map(str, grades))
print(grades_sentence1)

grades_sentence2 = ', '.join([ str(grade) for grade in grades ])
print(grades_sentence2)
````

Converting a list of numbers into a string using different separators.

````python
numbers = [4, 8, 5, 3, 6, 5]
print(' / '.join(map(str, numbers)))
print('_'.join(map(str, numbers)))
print(' - '.join(map(str, numbers)))
````

Converting a string into a list (separator is a space).

````python
sentence = 'Whatever sentence you want to convert into a list'
print(sentence)
print(sentence.split())
print(sentence.split(' '))
````

Converting a string into a list using different separators.

````python
sentence = 'This sentence / should, have commas, colons: and some / signs'
print(sentence)
print(sentence.split(' '))
print(sentence.split(' / '))
print(sentence.split(': '))
print(sentence.split(', '))
````

An example with dates provided as a string. The date string is converted into a list and additional strings are added.

````python
date_string = '2014-10-10|2015-12-30|2016-02-04|2017-05-03'
extra_dates = ['2018-11-08', '2019-07-08']

dates = date_string.split('|')
print(dates)

dates.extend(extra_dates)
print(dates)

joined_dates = '|'.join(dates)
print(joined_dates)
````

And here we attach the additional list as a string to the existing date string. Examine how everything works.

````python
date_string = '2014-10-10|2015-12-30|2016-02-04|2017-05-03'
extra_dates = ['2018-11-08', '2019-07-08']

joined_dates = date_string + '|' + '|'.join(extra_dates)
print(joined_dates)
````

This example shows how to convert a string of numbers into a list of numbers in two ways.

````python
numbers_as_text = '2/1/6/5/2/7/4'

numbers = numbers_as_text.split('/')
print(numbers)

numbers1 = list(map(int, numbers))
print(numbers1)

numbers2 = [int(num) for num in numbers]
print(numbers2)
````

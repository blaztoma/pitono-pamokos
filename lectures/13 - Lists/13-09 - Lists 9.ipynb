# Lists (`lists`) (Part 9)

## List transformation (`list comprehension`)

## Examples

```python
initial = [1, 2, 3, 4, 5, 6]
new = []

for el in initial:
    new.append(el)

print(initial)
print(new)
```

```python
initial = [1, 2, 3, 4, 5, 6]

new = [ el for el in initial ]

print(initial)
print(new)
```

```python
initial = [1, 2, 3, 4, 5, 6]
modified = []

for el in initial:
    modified.append( el * 2 )

print(initial)
print(modified)
```

```python
initial = [1, 2, 3, 4, 5, 6]

modified = [ el * 2 for el in initial ]

print(initial)
print(modified)
```

```python
letters = ['a', 'b', 'c']
modified = [ letter * 5 for letter in letters ]
print(letters)
print(modified)
```

```python
numbers = [5, 6, 7, 8, 2, 9]
print(numbers)
squares = [ n * n for n in numbers ]
print(squares)
```

```python
words = ['bird', 'apple', 'book', 'shop']
print('words:', words)
letter_counts = [ len(word) for word in words ]
print('letter counts:', letter_counts)
print('total letters:', sum(letter_counts))
```

```python
initial = [2, 1, 7, -2, -6, 7, -2, -5, 5, 6]
print('initial:', initial)

positives = []

for n in initial:
    if n > 0:
        positives.append(n)

print('selected positives:', positives)
```

```python
initial = [2, 1, 7, -2, -6, 7, -2, -5, 5, 6]
print('initial:', initial)

positives = [ n for n in initial if n > 0 ]
print('selected positives:', positives)

negatives = [ n for n in initial if n < 0 ]
print('selected negatives:', negatives)
```

```python
initial = ['Jonas', 'Ugnė', 'Petras', 'Ona', 'Gintarė', 'Povilas']
print(initial)
selected = [ name for name in initial if len(name) > 5 ]
print(selected)
```

```python
numbers = [3, 4, 8, 9, 2]
modified = [ 'even' if n % 2 == 0 else 'odd' for n in numbers ]
print(numbers)
print(modified)
```

```python
names = ['Jonas', 'Petras', 'Ugnė', 'Paulina']
print(names)

new = [ name[0].upper() for name in names ]
print(new)

new = [ name[0].upper() + name[1:] for name in names ]
print(new)
```

```python
names = ['Jonas', 'Petras', 'Ugnė', 'Paulina']
print(names)
new = [ name if name.startswith('P') else '---' for name in names ]
print(new)
```

```python
names = ['Jonas', 'Petras', None, 'Paulina', None]
print(names)

new = [ name if name != None else '' for name in names ]
print(new)

new = [ name if name is not None else '' for name in names ]
print(new)
```

```python
names = ['Jonas', 'Petras', None, 'Paulina', None]
print(names)
new = [ name for name in names if name is not None ]
print(new)
```

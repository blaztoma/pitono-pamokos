# Lists

## What are lists?
When programming, you often need to work with collections of data. Python is one of the most convenient languages for such things. A list is one of Python’s built-in data structures that lets you work with a collection of data in a sequential order.

Suppose we want to create a list of students:
* Tomas
* Rimas
* Rima
* Sigita

In a Python program, we can create a variable called `students` and assign all these names to it:

```python
students = ["Tomas", "Rimas", "Rima", "Sigita"]
print(students)
```

Did you notice these list-related things:
* a list starts with `[` and ends with `]`;
* list elements are separated by commas (no other separators work);
* after a comma, a space is usually added to make the code easier to read (but it will work without it too).

## Types of list elements
A question may arise: what can we put into a list? Well—many things! You saw an example with strings, but a list can contain elements of any Python type.

Here is what an `int` list looks like. Let these be the heights of student basketball players:

```python
student_heights = [180, 215, 205, 195]
print(student_heights)
```

Unlike in some other programming languages, we can store different types in one list. Let’s combine students’ names and their heights into a single list:

```python
students = ["Tomas", 180, "Rimas", 215, "Rima", 205, "Sigita", 195]
print(students)
```

## Empty lists
Most often, data is not written directly into the program. Data is stored in separate files, databases, internet servers. Until it is loaded, we can define an empty list that contains no elements.

```python
students = []
```

## List methods
A list in Python is an Object created from a corresponding class. Don’t worry if you don’t know this yet—we’ll learn it much later. The point is that a list has operations called methods. These operations let us manage list elements. By the way, we’ll use the terms `method` and `operation` as synonyms here, but if you want to be completely precise—use the term `method`.

### Adding elements using the append() method and plus
The `append()` operation adds an element to the end of a list. Let’s say we have the initial list of student names (Tomas, Rimas, Rima, Sigita) and we want to add a new student named Petras. We would do it like this:

```python
students = ["Tomas", "Rimas", "Rima", "Sigita"]
students.append("Petras")
print(students)
```

Now we can discuss the syntax. To use a list operation, you first write the list variable name (`students` in this case). Then you add a dot and write the method name. After that come parentheses, where we can provide the required data.

Now imagine you need to add not just one student, but a whole group. It would be very annoying to do that with `append()`—you’d have to write a lot. Luckily, we can add another list to a list using the plus (`+`) operation:

```python
students = ["Tomas", "Rimas", "Rima", "Sigita"]
students += ["Petras", "Antanas", "Nijolė"]
print(students)
```

Here we added an entire list, but in the same way we can also add just a single element. Just don’t forget to use list brackets (e.g., `students += ["Marytė"]`).

### Getting list elements using indices
Python numbers our list elements and we can access them using those numbers. For example, Python will number our initial student list like this:

| Element | Index |
|-----------|---------|
| Tomas | 0 |
| Rimas | 1 |
| Rima | 2 |
| Sigita | 3 |

Interesting! As you can see, Python numbers people and other elements starting from `0`. That’s just how it is—many languages do this for some reason...

So, let’s try printing the second element of the student list:


```python
students = ["Tomas", "Rimas", "Rima", "Sigita"]
print(students[1])
```

Did you notice that to print the second element we had to specify index `1`? Yes—because elements are numbered (indexed) starting from `0`.

Because of this (and other things), sometimes we might make a mistake. Let’s see how Python reacts:

```python
students = ["Tomas", "Rimas", "Rima", "Sigita"]
print(students[4])
```

You can see the error type is `IndexError`, and the description: `list out of range`. When we see that, we already know where the problem is.

This time, while experimenting, we exceeded the list boundary. But what would happen if we crossed the boundary the other way—i.e., we chose a negative index. Should it be the same `IndexError`!?

```python
students = ["Tomas", "Rimas", "Rima", "Sigita"]
print(students[-1])
```

Turns out—not at all! Python is a strange creature—it took the last element. It turns out that in Python we can index in reverse order too :) So the indices of our student list will be:

| Element | Index | Negative index |
|-----------|---------|-------------------|
| Tomas | 0 | -4 |
| Rimas | 1 | -3 |
| Rima | 2 | -2 |
| Sigita | 3 | -1 |

### Changing list elements
Changing a list element is simple. You just select the element to change using its index and assign a new value:

```python
students = ["Tomas", "Rimas", "Rima", "Sigita"]
students[1] = "Antanas"
print(students)
```

### Removing a list element
Imagine a big tragedy happened and student Tomas dropped out of the university after failing the math exam! Then we can remove him from the list using the `remove()` method:

```python
students = ["Tomas", "Rimas", "Rima", "Sigita"]
students.remove("Tomas")
print(students)
```

And what if there are two Tomases?

```python
students = ["Tomas", "Rimas", "Rima", "Tomas"]
students.remove("Tomas")
print(students)
```

Aha—it removed only the first Tomas. Want to remove the second one without removing the first? Then you’ll have to use one of the many list-handling functions: `del()`:

```python
students = ["Tomas", "Rimas", "Rima", "Tomas"]
del(students[-1])
print(students)
```

See—here we used an index for removal. There are more convenient operations—we’ll get to them.

## Two-dimensional lists

We already had a list of students and their heights. That is enough to manage student heights:

```python
students = ["Tomas", 180, "Rimas", 215, "Rima", 205, "Sigita", 195]
print(students[5])
```

But imagine that if we want to get the third student’s height, we have to use index `5`. That’s very inconvenient. It’s much better to have students and their heights aligned by the same index. That way we avoid mistakes due to inattention.

In Python, as in many other languages, this is done like this:

```python
students = [["Tomas", 180], ["Rimas", 215], ["Rima", 205], ["Sigita", 195]]
print(students[2], students[3])
```

See—much simpler :) What we did is organize each student’s data as a separate list. And each of those small lists is an element of the main list.

### Getting and changing elements of a 2D list

You can think of a two-dimensional list like a spreadsheet table.

| Indices | 0 | 1 |
|---|---|---|
| 0 | Tomas | 180 |
| 1 | Rimas | 215 |
| 2 | Rima | 205 |
| 3 | Sigita | 195 |

We can access elements using double indices:

```python
students = [["Tomas", 180], ["Rimas", 215], ["Rima", 205], ["Sigita", 195]]
print(students[0][0], students[0][1])
print(students[1][0], students[1][1])
print(students[2][0], students[2][1])
print(students[3][0], students[3][1])
```

Simple, right? The first index indicates the row, and the second indicates the column. If you ever mix them up—remember the spreadsheet :)

Now let’s change an element. This time we’ll replace student Rima with Marytė:

```python
students = [["Tomas", 180], ["Rimas", 215], ["Rima", 205], ["Sigita", 195]]
students[2][0] = "Marytė"
students[2][1] = 175
print(students)
```

This time we changed one element at a time, but if we need to replace the whole inner-list element, we can do it like this:

```python
students = [["Tomas", 180], ["Rimas", 215], ["Rima", 205], ["Sigita", 195]]
students[2] = ["Marytė", 175]
print(students)
```
